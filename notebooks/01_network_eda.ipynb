Цель: понять структуру данных, качество, баланс классов и различия признаков между нормой и атаками. Запускать из корня проекта (`jupyter notebook`).

```python
import sys; sys.path.insert(0, '..')
import pandas as pd, numpy as np, seaborn as sns, matplotlib.pyplot as plt
import config
from src.network.preprocess import load_network_csv, clean_network_data

raw = load_network_csv(config.NETWORK_DATA_PATH, require_label=True)
raw.head()
```

```python
df, quality = clean_network_data(raw)
quality
```

```python
# Баланс классов: норма сильно преобладает → Accuracy будет обманчиво высокой
df['Label'].value_counts().plot.barh(); plt.title('Распределение меток'); plt.show()
```

```python
# Как признаки различаются по классам (медианы)
df.groupby('Label')[config.NETWORK_FEATURES].median().T
```

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 4))
sns.boxplot(data=df, x='Label', y='Total Fwd Packets', ax=axes[0]); axes[0].set_yscale('log')
sns.boxplot(data=df, x='Label', y='Flow Packets/s', ax=axes[1]); axes[1].set_yscale('symlog')
for ax in axes: ax.tick_params(axis='x', rotation=30)
plt.tight_layout(); plt.show()
```

```python
# Корреляции признаков: сильно коррелирующие признаки дублируют информацию
plt.figure(figsize=(9, 7))
sns.heatmap(df[config.NETWORK_FEATURES].corr(), cmap='coolwarm', center=0)
plt.show()
```
